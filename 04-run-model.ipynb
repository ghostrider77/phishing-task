```python
from pathlib import Path

import pandas as pd

from predict import predict
```

```python
FOLDER = Path("data")
PATH = FOLDER / "test.csv"
```

```python
df = pd.read_csv(PATH)
```

```python
def get_prediction_probability(url: str) -> float:
    result = predict(url)
    return result["probability"]
```

```python
df["probability"] = df["url"].apply(get_prediction_probability)
```

```python
df.to_csv("predictions.csv", columns=["url", "probability"])
```
